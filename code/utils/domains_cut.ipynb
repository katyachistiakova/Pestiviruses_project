```python
from Bio import SeqIO
from Bio.SeqRecord import SeqRecord


def extract_hmmer_domains_domtblout(
    fasta_file,
    domtblout_file,
    output_file
):
    seqs = SeqIO.to_dict(SeqIO.parse(fasta_file, "fasta"))
    extracted = []

    with open(domtblout_file) as f:
        for line in f:
            if not line.strip() or line.startswith("#"):
                continue

            parts = line.split()
            seq_id = parts[3]
            env_from = int(parts[-4])
            env_to   = int(parts[-3])

            if seq_id not in seqs:
                print(f"{seq_id} not in FASTA")
                continue

            nt_start = (env_from - 1) 
            nt_end   = env_to 

            subseq = seqs[seq_id].seq[nt_start:nt_end]

            record = SeqRecord(
                subseq,
                id=f"{seq_id}_{env_from}_{env_to}",
                description="RdRp_env_coords_nt"
            )

            extracted.append(record)

    SeqIO.write(extracted, output_file, "fasta")
    print(f"{len(extracted)} domains saved")

extract_hmmer_domains_domtblout(
    fasta_file=r"C:\Users\2slon\OneDrive\Рабочий стол\5sem\pestiviruses\final\small\border_desease\aa_swine_out.fasta",
    domtblout_file=r"C:\Users\2slon\OneDrive\Рабочий стол\5sem\pestiviruses\final\classical_swine_fever_virus\swine\swine_dom.tbl",
    output_file=r"C:\Users\2slon\OneDrive\Рабочий стол\5sem\pestiviruses\final\small\border_desease\swine_out_RDRP")

```
